[문제 6] 0부터 시작하여 n까지의 숫자 중에서 홀수만 생성하는 제너레이터 odd_numbers를 작성하세요.

def odd_numbers(n):
    for number in range(n + 1):
        if number % 2 != 0:
            yield number

# 1. n=10을 인수로 제너레이터 객체 생성
odd_gen = odd_numbers(10)
print("--- n=10까지의 홀수 ---")

# 2. next()를 사용하여 값을 하나씩 꺼냄
print(next(odd_gen)) 
print(next(odd_gen)) 

# 3. for 루프를 사용하여 나머지 값 출력
for odd in odd_gen:
    print(odd)

[문제 10] 팩토리얼을 계산하는 재귀 함수 factorial을 작성하세요.

def factorial(n: int) -> int:
    
    if n == 0 or n == 1:
        return 1

    else:
        return n * factorial(n - 1)

[문제 12] 두 문자열을 받아서 긴 문자열을 반환하는 함수 longer_string을 작성하세요. 만약 두 문자열의 길이가 같다면, 첫 번째 문자열을 반환하세요.

def longer_string(str1: str, str2: str) -> str:

    len1 = len(str1)
    len2 = len(str2)

    if len2 > len1:
        return str2
    else:
        return str1
